## Part 4 - Visualization - Correlation Heatmap

### Task 4.1

Create a correlation dataframe for the dataset (all columns against all columns).

Use seaborn to plot 2 correlation heatmaps:
1. Where values are going from -1 to 1.
2. Where values are going from 0 to 1.

````python
# Correlation matrix (all columns vs all columns)
corr = df.corr(numeric_only=True)

# Heatmap 1: values from -1 to 1
plt.figure(figsize=(12, 10))
sns.heatmap(corr, vmin=-1, vmax=1, annot=False)
plt.title('Correlation Heatmap (-1 to 1)')
plt.show()

# Heatmap 2: values from 0 to 1 (absolute correlation)
corr_abs = corr.abs()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_abs, vmin=0, vmax=1, annot=False)
plt.title('Correlation Heatmap (0 to 1) - Absolute Correlation')
plt.show()
````

### Task 4.2

Look at the heatmap, choose the pair of columns with the highest correlation between them,
and draw graph to show the linear relation between those 2 columns.
You have to choose the correct graph type to show that!

````python
# נשתמש במטריצת קורלציה בערכים מוחלטים כדי למצוא את הזוג הכי חזק
corr_abs = df.corr(numeric_only=True).abs()

# כדי לא לבחור את האלכסון (קורלציה של עמודה עם עצמה = 1)
np.fill_diagonal(corr_abs.values, 0)

# למצוא את הזוג עם הערך הכי גבוה
max_idx = np.unravel_index(np.argmax(corr_abs.values), corr_abs.shape)
col_x = corr_abs.index[max_idx[0]]
col_y = corr_abs.columns[max_idx[1]]
max_corr = corr_abs.iloc[max_idx]

print(f"Highest correlation pair: {col_x} & {col_y} (abs corr = {max_corr:.4f})")

# גרף שמראה קשר לינארי: scatter + קו רגרסיה
plt.figure(figsize=(8, 6))
sns.regplot(x=df[col_x], y=df[col_y], scatter_kws={'alpha': 0.5})
plt.title(f'Linear Relation: {col_x} vs {col_y} (abs corr={max_corr:.2f})')
plt.xlabel(col_x)
plt.ylabel(col_y)
plt.show()
````
